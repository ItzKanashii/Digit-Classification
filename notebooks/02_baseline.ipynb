# 02 · Baseline classifiers

This notebook sets a reference point with two models that use flattened pixel values: logistic regression and a multilayer perceptron (MLP). Both are evaluated on the same held-out test split.


```python
from src.data import load_emnist
from src.preprocessing import preprocess_images
from src.models import create_logistic_regression, create_mlp
```

## Prepare the data

Load EMNIST and scale pixel values to the range 0–1. Flattening turns each 28 × 28 image into a feature vector for these classifiers.


```python
X_train, y_train, writers_train, X_test, y_test, writers_test = load_emnist()

X_train = preprocess_images(X_train)
X_test = preprocess_images(X_test)
```

```python
print(X_train.shape)
print(X_train.dtype)
print(X_train.min())
print(X_train.max())
```

```python
X_train_flat = X_train.reshape(len(X_train), -1)
X_test_flat = X_test.reshape(len(X_test), -1)
```

## Logistic regression

Fit a linear classifier, then measure its accuracy on the test images.


```python
model = create_logistic_regression()

model.fit(X_train_flat, y_train)
```

```python
accuracy = model.score(X_test_flat, y_test)

print(f"Test accuracy: {accuracy:.4f}")
```

## Multilayer perceptron

The MLP adds hidden layers so it can learn nonlinear combinations of pixels.


```python
mlp = create_mlp()

mlp.fit(X_train_flat, y_train)
```

```python
mlp_accuracy = mlp.score(X_test_flat, y_test)

print(f"MLP test accuracy: {mlp_accuracy:.4f}")
```

## Results

In the saved run, logistic regression reached **94.18%** test accuracy and the MLP reached **99.05%**. The MLP is the stronger baseline in this comparison.
